
### 2) Linear vs Logistic Regression 📈➡️📊


```

# TODO: Residual analysis and logistic cross-entropy demonstration

import numpy as np
from sklearn.datasets import make_regression, make_classification
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error

# Linear regression
X_reg, y_reg = make_regression(n_samples=200, n_features=1, noise=10, random_state=42)
lin_model = LinearRegression().fit(X_reg, y_reg)
y_pred = lin_model.predict(X_reg)

print("MSE:", mean_squared_error(y_reg, y_pred))

# Logistic regression
X_clf, y_clf = make_classification(n_samples=200, n_features=2, random_state=42)
log_model = LogisticRegression().fit(X_clf, y_clf)
print("Logistic model trained successfully")
```


### 3) KNN Hyperparameters & Decision Trees 🌳


```

# TODO: KNN hyperparameter sweep and Decision Tree depth comparison

from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

accuracies = []

for k in range(1, 21):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_clf, y_clf)
    accuracies.append(accuracy_score(y_clf, knn.predict(X_clf)))

print("KNN accuracy range:", min(accuracies), "to", max(accuracies))

tree = DecisionTreeClassifier(max_depth=3, random_state=42)
tree.fit(X_clf, y_clf)
print("Decision Tree depth 3 accuracy:", accuracy_score(y_clf, tree.predict(X_clf)))
```
